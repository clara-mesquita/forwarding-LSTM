```python
import json

# Caminho do arquivo
file_path = "./bbr/evaluation.json"

# Carregar o JSON
with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Filtrar chaves com NRMSE <= 0.15
filtered = {
    k: v for k, v in data.items()
    if v.get("NRMSE", float("inf")) <= 0.15
}

# Printar resultados
print(f"Quantidade de links com NRMSE <= 0.15: {len(filtered)}\n")

for key, metrics in filtered.items():
    link = key.split("_")[0]
    nrmse = metrics["NRMSE"]

    print(f"Chave completa: {key}")
    print(f"Link: {link}")
    print(f"NRMSE: {nrmse:.4f}")
    print("-" * 40)
```

```python
import json
import os
import shutil

# Paths
evaluation_path = "./bbr/evaluation.json"
source_dir = "../datasets/treated/bbr"
target_dir = "../datasets/amostras"

# Criar pasta de destino se não existir
os.makedirs(target_dir, exist_ok=True)

# Carregar evaluation.json
with open(evaluation_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Filtrar chaves com NRMSE <= 0.15
selected_keys = [
    k for k, v in data.items()
    if v.get("NRMSE", float("inf")) <= 0.15
]

print(f"Quantidade de arquivos a copiar: {len(selected_keys)}\n")

# Copiar arquivos
copied = 0
not_found = []

for key in selected_keys:
    key = key + ".csv"
    src_file = os.path.join(source_dir, key)
    dst_file = os.path.join(target_dir, key)

    if os.path.isfile(src_file):
        shutil.copy2(src_file, dst_file)
        copied += 1
        print(f"Copiado: {key}")
    else:
        not_found.append(key)

print("\nResumo:")
print(f"Arquivos copiados: {copied}")
print(f"Arquivos não encontrados: {len(not_found)}")

if not_found:
    print("\nArquivos não encontrados:")
    for k in not_found:
        print(f"- {k}")
```

```python
import json

file_path = "./bbr/evaluation.json"

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# 1. Filtrar por NRMSE <= 0.15
candidates = {
    k: v for k, v in data.items()
    if v.get("NRMSE", float("inf")) <= 0.15
}

# 2. Ordenar candidatos por NRMSE crescente
sorted_candidates = sorted(
    candidates.items(),
    key=lambda x: x[1]["NRMSE"]
)

selected = {}
used_states = set()

# 3. Selecionar links sem repetir estados
for key, metrics in sorted_candidates:
    link = key.split("_")[0]      # ex: "am-mt"
    states = set(link.split("-")) # {"am", "mt"}

    # Verifica conflito de estados
    if states & used_states:
        continue

    selected[key] = metrics
    used_states.update(states)

# 4. Printar resultados
print(f"Quantidade final de links (sem conflito de estados): {len(selected)}\n")

for key, metrics in selected.items():
    link = key.split("_")[0]
    print(f"Chave completa: {key}")
    print(f"Link: {link}")
    print(f"NRMSE: {metrics['NRMSE']:.4f}")
    print("-" * 40)
```
